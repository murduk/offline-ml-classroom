# NumPy and pandas — local data

This notebook reads a CSV bundled in the Docker image. No downloads are needed.

```python
import numpy as np
import pandas as pd
from pathlib import Path

data = pd.read_csv(Path("../datasets/housing-sample.csv"))
data.head()
```

```python
print(data.describe())
features = data[["area_m2", "bedrooms", "age_years"]].to_numpy()
print("Feature array shape:", features.shape)
print("Mean area:", np.mean(features[:, 0]))
```

```python
import matplotlib.pyplot as plt
plt.scatter(data["area_m2"], data["price_usd"])
plt.xlabel("Area (m²)")
plt.ylabel("Price (USD)")
plt.title("Synthetic housing data")
plt.show()
```
