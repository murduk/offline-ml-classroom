# Clustering — synthetic groups

Generate points locally and cluster them. Change the number of clusters and compare the results.

```python
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import matplotlib.pyplot as plt

X, true_groups = make_blobs(n_samples=300, centers=3, cluster_std=0.8, random_state=42)
```

```python
model = KMeans(n_clusters=3, n_init=10, random_state=42)
labels = model.fit_predict(X)
print("Silhouette score:", round(silhouette_score(X, labels), 3))
```

```python
plt.scatter(X[:, 0], X[:, 1], c=labels, s=20)
plt.scatter(model.cluster_centers_[:, 0], model.cluster_centers_[:, 1], marker="X", s=150, c="red")
plt.title("K-means clusters")
plt.show()
```
