# Regression — predicting synthetic prices

Train/test splitting keeps the test set separate. These synthetic records are only for teaching.

```python
from pathlib import Path
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_absolute_error, r2_score

data = pd.read_csv(Path("../datasets/housing-sample.csv"))
X = data[["area_m2", "bedrooms", "age_years"]]
y = data["price_usd"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
```

```python
model = make_pipeline(StandardScaler(), Ridge(alpha=1.0))
model.fit(X_train, y_train)
predictions = model.predict(X_test)
print("MAE (USD):", round(mean_absolute_error(y_test, predictions)))
print("R²:", round(r2_score(y_test, predictions), 3))
```

```python
import matplotlib.pyplot as plt
plt.scatter(y_test, predictions)
low, high = min(y_test.min(), predictions.min()), max(y_test.max(), predictions.max())
plt.plot([low, high], [low, high], "k--")
plt.xlabel("Actual price")
plt.ylabel("Predicted price")
plt.show()
```
